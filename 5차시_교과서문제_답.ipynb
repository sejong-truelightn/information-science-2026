## 문제 1 &nbsp;&nbsp;`상태 공간의 배제`

현재 사용할 수 있는 동전은 10원, 50원, 100원, 500원으로 구성되어 있다.
`n` 원을 지불할 때 최소 동전의 개수를 구해 보자.

같은 문제를 **① 전체 탐색 → ② 분기 한정법 → ③ 탐욕법** 순서로 풀고 비교한다.

### ① 전체 탐색

동전을 하나 고를 때마다 4갈래로 갈라진다. 상태는 **(지금까지 낸 금액 `paid`, 낸 동전 수 `count`)** 이다.

①은 `n` 원을 넘은 경우다. 더 내려가도 `n` 원이 될 수 없으므로 백트랙한다.
②에서 `n` 원을 정확히 맞추면 지금까지의 최소 개수와 비교한다.

전체 탐색은 `n` 이 조금만 커져도 끝나지 않는다. 그래서 `n = 450` 으로 실행한다.

**실행 결과**
```
n = 450
최소 동전 개수 : 5
함수 호출 횟수 : 4648653
```

````python
coins = [500, 100, 50, 10]
min_count = 987654321      # 지금까지 찾은 최소 개수
calls = 0                  # 함수가 몇 번 불렸는지


def pay(paid, count):      # paid = 지금까지 낸 금액, count = 낸 동전 수
    global min_count, calls
    calls = calls + 1
    if paid > n:           # ①
        return
    if paid == n:          # ②
        min_count = min(min_count, count)
        return
    for coin in coins:
        pay(paid + coin, count + 1)


n = 450
pay(0, 0)

print('n =', n)
print('최소 동전 개수 :', min_count)
print('함수 호출 횟수 :', calls)
````

#### 단계별로 쌓아 보기 &nbsp;&nbsp;<span style="color:#6b4bc4">위 코드를 4단계로 나누기</span>

위 코드를 한 줄씩 덧붙이며 실행해 본다. 1~3단계는 `n` 을 작게 두고 출력으로 흐름을 직접 확인한다.

**1단계 · 상태와 4갈래**

상태는 (지금까지 낸 금액 `paid`, 낸 동전 수 `count`) 이다. 매번 동전 4개 중 하나를 더 낸다. 들여쓰기가 곧 트리 모양이다.
①이 없으면 금액이 계속 커져서 재귀가 끝나지 않는다.

**실행 결과**
```
 0 0
   500 1
   초과
   100 1
   초과
   50 1
   초과
   10 1
     510 2
     초과
     110 2
     초과
     60 2
     초과
     20 2
     정답 2
```

````python
coins = [500, 100, 50, 10]


def pay(paid, count):      # paid = 지금까지 낸 금액, count = 낸 동전 수
    print('  ' * count, paid, count)
    if paid > n:           # ①
        print('  ' * count, '초과')
        return
    if paid == n:          # ②
        print('  ' * count, '정답', count)
        return
    for coin in coins:
        pay(paid + coin, count + 1)


n = 20
pay(0, 0)
````

**2단계 · 답이 여러 개 나온다**

`n = 60` 이면 정답이 세 번 나온다. 50 → 10 과 10 → 50 은 **순서만 다른 같은 조합**인데도 따로 센다.
답이 여러 개이므로 그중 **가장 적은 것**을 기억해 둬야 한다.

**실행 결과**
```
정답 발견! 동전 2 개
정답 발견! 동전 2 개
정답 발견! 동전 6 개
```

````python
coins = [500, 100, 50, 10]


def pay(paid, count):
    if paid > n:           # ①
        return
    if paid == n:          # ②
        print('정답 발견! 동전 %d 개' % count)
        return
    for coin in coins:
        pay(paid + coin, count + 1)


n = 60
pay(0, 0)
````

**3단계 · `min_count` 로 최솟값 기억하기**

정확히 `n` 원이 될 때마다 `min(min_count, count)` 로 더 작은 쪽을 남긴다.
처음 값을 987654321 처럼 크게 두어야 첫 답이 무조건 들어간다. 함수 안에서 대입하므로 `global` 이 필요하다.

**실행 결과**
```
정답 발견! 동전 2 개 → min_count = 2
정답 발견! 동전 2 개 → min_count = 2
정답 발견! 동전 6 개 → min_count = 2
최소 동전 개수 : 2
```

````python
coins = [500, 100, 50, 10]
min_count = 987654321      # 지금까지 찾은 최소 개수


def pay(paid, count):
    global min_count
    if paid > n:           # ①
        return
    if paid == n:          # ②
        min_count = min(min_count, count)
        print('정답 발견! 동전 %d 개 → min_count = %d' % (count, min_count))
        return
    for coin in coins:
        pay(paid + coin, count + 1)


n = 60
pay(0, 0)
print('최소 동전 개수 :', min_count)
````

**4단계 · `calls` 로 탐색량 재기**

트리 출력을 지우고 `calls = calls + 1` 을 넣으면 맨 위의 전체 탐색 코드가 된다.

```
  n | 최소 개수 | 호출 횟수
 60 |         2 |        29
450 |         5 |   4648653
```

450원에 465만 번이다. 5개짜리 답을 찾은 뒤에도 6개, 7개 … 쓰는 갈래를 끝까지 다 가 본다.
이 낭비를 자르는 것이 다음의 **분기 한정법**이다.

### ② 분기 한정법

`min_count` 개로 `n` 원을 만드는 방법을 이미 찾았다면, 동전을 `min_count` 개 이상 쓴 갈래는 볼 필요가 없다.
전체 탐색 코드에 ③ **한 줄**을 더한 것이 전부다.

`coins` 를 **큰 동전부터** 넣어 둔 것도 중요하다. 처음 찾는 답이 500원부터 채운 9개라서, 그 뒤로 10개 이상 쓰는 갈래가 전부 잘린다.

**실행 결과**
```
n = 1780
최소 동전 개수 : 9
함수 호출 횟수 : 290697
```

````python
coins = [500, 100, 50, 10]
min_count = 987654321
calls = 0


def pay(paid, count):
    global min_count, calls
    calls = calls + 1
    if paid > n:
        return
    if count >= min_count: # ③ 이미 min_count개로 만들었다면 더 볼 필요가 없다
        return
    if paid == n:
        min_count = min(min_count, count)
        return
    for coin in coins:
        pay(paid + coin, count + 1)


n = 1780
pay(0, 0)

print('n =', n)
print('최소 동전 개수 :', min_count)
print('함수 호출 횟수 :', calls)
````

### 호출 횟수 비교 &nbsp;&nbsp;<span style="color:#6b4bc4">전체 탐색 vs 분기 한정법</span>

두 함수는 ③ 한 줄만 다르다. 금액을 50원씩 늘려 가며 호출 횟수를 세어 본다.

**실행 결과**
```
  n | 최소 개수 | 전체 탐색 호출 | 분기 한정 호출
100 |         1 |            101 |              5
150 |         2 |            481 |             17
200 |         2 |           2221 |             17
250 |         3 |          10245 |             53
300 |         3 |          47265 |             53
350 |         4 |         218149 |            161
400 |         4 |        1007009 |            161
450 |         5 |        4648653 |            485
```

````python
coins = [500, 100, 50, 10]


def count_calls(n, bound):
    min_count = 987654321
    calls = 0

    def pay(paid, count):
        nonlocal min_count, calls
        calls = calls + 1
        if paid > n:
            return
        if bound and count >= min_count:   # bound=True 일 때만 분기 한정
            return
        if paid == n:
            min_count = min(min_count, count)
            return
        for coin in coins:
            pay(paid + coin, count + 1)

    pay(0, 0)
    return min_count, calls


print('  n | 최소 개수 | 전체 탐색 호출 | 분기 한정 호출')
for n in range(100, 451, 50):
    min_count, full = count_calls(n, False)
    _, pruned = count_calls(n, True)
    print('%3d | %9d | %14d | %14d' % (n, min_count, full, pruned))
````

#### 동전 순서를 바꾸면 &nbsp;&nbsp;<span style="color:#6b4bc4">분기 한정의 효과가 달라진다</span>

분기 한정은 &lsquo;지금까지의 최솟값&rsquo;을 기준으로 자른다. 기준이 일찍 작아질수록 많이 잘린다.
작은 동전부터 고르면 처음 찾는 답이 **10원 178개**라서 기준이 한참 동안 느슨하다.

**실행 결과**
```
n = 1780
큰 동전부터   : 9 개, 호출 290697 번
작은 동전부터 : 9 개, 호출 9098965 번
```

````python
def count_calls(n, coins):
    min_count = 987654321
    calls = 0

    def pay(paid, count):
        nonlocal min_count, calls
        calls = calls + 1
        if paid > n:
            return
        if count >= min_count:
            return
        if paid == n:
            min_count = min(min_count, count)
            return
        for coin in coins:
            pay(paid + coin, count + 1)

    pay(0, 0)
    return min_count, calls


n = 1780

print('n =', n)
print('큰 동전부터   : %d 개, 호출 %d 번' % count_calls(n, [500, 100, 50, 10]))
print('작은 동전부터 : %d 개, 호출 %d 번' % count_calls(n, [10, 50, 100, 500]))
````

### ③ 탐욕법

가장 큰 동전부터 낼 수 있는 만큼 낸다. 갈래를 **하나만** 남기고 나머지는 전부 버리는 셈이다.
반복은 동전 종류 수(4번)뿐이라 `n` 이 아무리 커도 바로 끝난다.

`//` 는 몫, `%` 는 나머지다. 1780원이면 500원 3개(1500원)를 내고 280원이 남는다.

**실행 결과**
```
n = 1780
500원 3개
100원 2개
50원 1개
10원 3개
최소 동전 개수 : 9
```

````python
coins = [500, 100, 50, 10]

n = 1780
rest = n                   # 아직 내야 하는 금액
count = 0

print('n =', n)
for coin in coins:         # 큰 동전부터
    k = rest // coin       # coin원짜리를 낼 수 있는 만큼
    rest = rest % coin     # 내고 남은 금액
    count = count + k
    print('%d원 %d개' % (coin, k))

print('최소 동전 개수 :', count)
````

#### 탐욕법이 틀리는 동전 &nbsp;&nbsp;<span style="color:#6b4bc4">500원 · 400원 · 100원</span>

10·50·100·500원은 큰 동전이 바로 아래 동전의 배수라서 탐욕법이 언제나 맞는다.
400원짜리 동전이 있다고 하면 800원에서 틀린다. 탐욕법은 500원을 먼저 내고 100원 3개를 더 내지만, 400원 2개면 된다.

**실행 결과**
```
n = 800 , 동전 [500, 400, 100]
탐욕법    : 4 개
분기 한정 : 2 개
```

````python
def greedy(n, coins):
    count = 0
    for coin in coins:
        count = count + n // coin
        n = n % coin
    return count


def branch_bound(n, coins):
    min_count = 987654321

    def pay(paid, count):
        nonlocal min_count
        if paid > n or count >= min_count:
            return
        if paid == n:
            min_count = count
            return
        for coin in coins:
            pay(paid + coin, count + 1)

    pay(0, 0)
    return min_count


coins = [500, 400, 100]
n = 800

print('n =', n, ', 동전', coins)
print('탐욕법    :', greedy(n, coins), '개')
print('분기 한정 :', branch_bound(n, coins), '개')
````

## 문제 2 &nbsp;&nbsp;`상태 공간의 배제`

100억 이하의 자연수 `n` 이 주어질 때 `n` 의 모든 약수의 합을 구해 보자.

같은 문제를 **① 전체 탐색 → ② 상태 공간의 배제** 순서로 풀고 비교한다.

### ① 전체 탐색

1차시에 푼 방법 그대로다. 1부터 `n` 까지 하나씩 나누어 본다.

**실행 결과**
```
n = 1000
약수의 합 : 2340
```

````python
n = 1000
divisor_sum = 0

for i in range(1, n + 1):
    if n % i == 0:
        divisor_sum = divisor_sum + i

print('n =', n)
print('약수의 합 :', divisor_sum)
````

### ② 상태 공간의 배제

약수는 짝으로 나온다. `i` 가 약수이면 `n // i` 도 약수다(1000 = 8 × 125).
짝 중 작은 쪽은 언제나 √n 이하이므로, ① `i * i <= n` 까지만 보고 큰 쪽은 `n // i` 로 바로 구한다.

②는 제곱수 처리다. 16 = 4 × 4 처럼 짝이 자기 자신이면 한 번만 더한다.
`i * i <= n` 으로 쓰면 `n ** (1/2)` 같은 실수 계산이 필요 없다.

**실행 결과**
```
n = 1000
약수의 합 : 2340
반복 횟수 : 31
```

````python
n = 1000
divisor_sum = 0

i = 1
while i * i <= n:          # ① √n 까지만
    if n % i == 0:
        divisor_sum = divisor_sum + i
        if i != n // i:    # ② 제곱수면 짝이 자기 자신이다
            divisor_sum = divisor_sum + n // i
    i = i + 1

print('n =', n)
print('약수의 합 :', divisor_sum)
print('반복 횟수 :', i - 1)
````

#### for 문으로 쓰기

`range` 의 끝값으로 √n 을 넣는다. `n ** 0.5` 는 실수이므로 `int()` 로 정수로 바꾼다.

**실행 결과**
```
n = 1000
약수의 합 : 2340
반복 횟수 : 31
```

````python
n = 1000
divisor_sum = 0

root = int(n ** 0.5)               # √n 을 정수로
for i in range(1, root + 1):       # ① √n 까지만
    if n % i == 0:
        divisor_sum = divisor_sum + i
        if i != n // i:            # ② 제곱수면 짝이 자기 자신이다
            divisor_sum = divisor_sum + n // i

print('n =', n)
print('약수의 합 :', divisor_sum)
print('반복 횟수 :', root)
````

#### 제곱수에서 짝을 한 번만 더하는 이유

`n = 36` 의 짝을 보면 다음과 같다.

| i | n // i |
|---|---|
| 1 | 36 |
| 2 | 18 |
| 3 | 12 |
| 4 | 9 |
| **6** | **6** ← 짝이 자기 자신 |

`i = 6` 일 때 짝도 6이다. 무조건 둘 다 더하면 6이 **두 번** 더해진다.

- 틀린 결과: 1+36+2+18+3+12+4+9+**6+6** = 97
- 맞는 결과: 1+2+3+4+6+9+12+18+36 = **91**

그래서 아래 줄에서 `i` 는 항상 더하고, 짝 `n // i` 는 **`i` 와 다를 때만** 더한다.

```python
if i != n // i:                        # 짝이 나와 다를 때만
    divisor_sum = divisor_sum + n // i # 짝을 더한다
```
