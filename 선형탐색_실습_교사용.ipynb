## 1. 교과서 코드 풀어보기 — 약수의 합 구하기

자연수 n이 주어질 때, n의 모든 약수의 합을 구해 보자.

탐색해야 하는 상태 공간은 `{1, 2, 3, ..., n}` 이다.

**주어진 데이터**: `n = 10`  →  **실행 결과**: `18`

아래 셀을 실행하고, 각 줄이 무슨 역할을 하는지 옆 사람과 이야기해보자.

```python
n = 10
ans = 0
for i in range(1, n+1):
    if (n % i) == 0:
        ans = ans + i
print(ans)
```

## 2. 순차 탐색 스스로 해보기 — 홀수 개수 구하기

주어진 데이터 목록에서 홀수의 개수를 구해 보자.

**주어진 데이터**: `data = [4, 7, 12, 3, 9, 18, 21, 6]`

**실행 결과**: `4`

```python
data = [4, 7, 12, 3, 9, 18, 21, 6]
ans = 0
for i in range(len(data)):
    if data[i] % 2 == 1:
        ans = ans + 1
print(ans)
```

## 3. 이분 탐색 직접 완성하기

정렬된 리스트 `arr`에서 특정 값 `target`이 있는 인덱스를 이분 탐색으로 찾아보자. (없으면 `-1`)


**주어진 데이터**: `arr = [2, 4, 6, 8, 10, 12, 14]`, `target = 10`

**실행 결과**: `4`

```python
arr = [2, 4, 6, 8, 10, 12, 14]
target = 10
low, high = 0, len(arr) - 1
result = -1

while low <= high:
    mid = (low + high) // 2
    if arr[mid] == target:
        result = mid
        break
    elif arr[mid] < target:
        low = mid + 1
    else:
        high = mid - 1

print(result)
```

## 4. 스택 실습 — 괄호 짝 검사

주어진 문자열에 있는 소괄호 `(` `)` 의 짝이 맞는지 **스택**으로 판별해보자.

여는 괄호는 스택에 넣고(push), 닫는 괄호를 만나면 스택에서 하나를 꺼낸다(pop). 짝이 맞지 않거나 끝까지 다 짝지어지지 않은 괄호가 남으면 `NO`이다.

**주어진 데이터**: `s = "(()())"`

**실행 결과**: `YES`

```python
s = "(()())"
stack = []
valid = True

for ch in s:
    if ch == '(':
        stack.append(ch)
    elif ch == ')':
        if not stack:
            valid = False
            break
        stack.pop()

if stack:
    valid = False

print("YES" if valid else "NO")
```
