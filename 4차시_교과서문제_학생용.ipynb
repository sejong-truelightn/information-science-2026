# 부록 · 개념 점검 ①  파이썬

다섯 문제 **전부**에 쓰이는 것들이다. 하나라도 흔들리면 답이 아니라 `None` 이나 `0` 이 나온다.

| 개념 | 이 노트북에서 걸리는 지점 |
|---|---|
| **`return` 과 `print` 의 차이** | 뼈대가 `print('...', count(n))` 형태다. 함수는 **값을 돌려줘야** 한다. 함수 안에서 출력하면 `None` 이 같이 찍힌다 |
| **함수 정의 · 매개변수** | 매개변수가 곧 **상태**다. &ldquo;무엇만 알면 나머지가 정해지는가&rdquo;를 매개변수로 적는다 |
| **`%` 나머지 연산** | `a % b == 0` 이면 나누어떨어진다. 문제 3·4·5의 판정이 전부 이것이다 |
| **`range(a, b)` 는 b를 포함하지 않는다** | `range(1, n+1)` 에서 **+1** 을 빠뜨리는 실수가 가장 흔하다 |
| **누적 변수** | `count = count + 1` — 세는 문제는 전부 이 꼴이다 |
| **불리언 반환** | `is_prime(x)` 는 `True`/`False` 를 돌려준다. 그래야 `if is_prime(x):` 로 쓸 수 있다 |
| **`global`** | 함수 **안에서 바깥 변수를 바꾸려면** 필요하다 (문제 2·5) |
| **재귀 — 종료 조건 + 자기 호출** | 종료 조건이 없으면 **영원히** 내려간다 (문제 1·2·5) |

### 가장 많이 나오는 실수

```python
# ✗ 함수가 값을 안 돌려준다
def count(rest):
    print(rest)      # 출력만 하고 끝

print(count(3))
# -> 3
#    None   <- 돌려준 값이 없다


# ✓ 값을 돌려준다
def count(rest):
    return 4

print(count(3))      # -> 4
```

> `pass` 는 &ldquo;아무것도 하지 않는다&rdquo;는 뜻이다.
> 뼈대를 그대로 실행하면 `None` 이나 `0` 이 나온다. **오류가 아니다.** 왜 그 값이 나왔는지부터 설명할 수 있어야 한다.

# 부록 · 개념 점검 ②  수학 용어와 만나게 될 오류

### 수학 용어 여섯 개

| 용어 | 뜻 | 쓰이는 곳 |
|---|---|---|
| **약수** | 나누어떨어지게 하는 수. `n % d == 0` | 문제 3 |
| **배수** | 약수의 반대편. 3의 배수 → `s % 3 == 0` | 문제 5 |
| **소수** | 1과 자기 자신 외에 약수가 없는, **1보다 큰** 수 | 문제 3 |
| **최대공약수** | 두 수의 공통 약수 중 가장 큰 것 | 문제 4 |
| **서로소** | 최대공약수가 **1**인 두 수 | 문제 4 |
| **순열 · 조합** | 순서를 **따지면** 순열, **안 따지면** 조합 | 문제 1 · 5 |

### 순열과 조합이 한 노트북에 같이 있다

| | 문제 1 **순열** | 문제 5 **조합** |
|---|---|---|
| **다르게 세나?** | `1+2` 와 `2+1` 은 **다르다** | `(1,3,6)` 과 `(3,1,6)` 은 **같다** |
| **코드의 차이** | 매번 **1부터** 다시 → `range(1, rest+1)` | **다음 것부터** → `range(start, N)` 과 `i+1` |

이 한 글자 차이가 답을 바꾼다. **문제를 읽고 어느 쪽인지 먼저 정하는 것**이 코드보다 앞선다.

### 만나게 될 오류 세 가지

| 화면에 나오는 것 | 원인 | 어느 문제 |
|---|---|---|
| **`RecursionError`** | 재귀에 **종료 조건이 없다**. 계단 문제에서 `c > n` 검사를 빠뜨리면 계단을 넘어서도 계속 올라간다 | 문제 1 · 2 · 5 |
| **`UnboundLocalError`** | `global` 없이 함수 안에서 바깥 변수를 바꾸려 했다 | 문제 5 |
| **`None`** 또는 **`0`** | 오류는 아니다. `return` 을 안 썼거나, **1부터 n까지 훑는 반복문을 안 썼다** | 전부 |

> ⚠️ **문제 3 · 4의 뼈대에는 `for` 문이 아예 없다.** `count = 0` 만 있고, 1부터 n까지 훑는 반복문은 직접 써야 한다.
> 함수만 채우고 끝내면 계속 `0` 이 나온다.
