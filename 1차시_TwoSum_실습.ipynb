## 문제
정수로 이루어진 리스트 `nums`와 목표값 `target`이 주어질 때, 리스트 안에서 **더해서 target이 되는 두 원소의 인덱스**를 찾아 반환하세요.
(정답이 딱 하나 존재한다고 가정합니다.)

**입력 예시**
```
nums = [2, 7, 11, 15]
target = 9
```
**출력 예시**
```
[0, 1]   # nums[0] + nums[1] = 2 + 7 = 9
```

## 조건
- 리스트의 길이 `n`은 최대 1,000,000까지 주어질 수 있습니다.
- 아래에서 직접 시간을 재보며 확인해봅시다.

## 미션
1. 가장 먼저 떠오르는 방법(이중 반복문)으로 풀고, 실행 시간을 측정해보세요.
2. 왜 느린지 시간 복잡도로 설명해보세요.
3. 이중 반복문 없이 더 빠르게 푸는 방법을 찾아 직접 구현하고, 시간을 비교해보세요.

## 0️⃣ 실험용 데이터 만들기
아래 셀을 실행해서 테스트 데이터를 준비하세요.

````python
import random
import time

random.seed(42)
n = 50_000  # 리스트 크기 (나중에 값을 바꿔서도 실험해보세요)
nums = random.sample(range(1, 5_000_000_000), n)  # 서로 다른 정수 n개
target = nums[-1] + nums[-2]  # 정답이 존재하도록 마지막 두 값의 합을 target으로 설정

print(f'리스트 길이: {len(nums)}')
print(f'target: {target}')
````

## 1️⃣ 처음 떠오르는 방법: 이중 반복문
모든 쌍 (i, j)을 하나하나 비교하며 합이 target인지 확인하는, 가장 직관적인 방법입니다.

````python
def two_sum_brute(nums, target):
   for i in range(len(nums)):
    for j in range(i+1, len(nums)):
      if nums[i]+nums[j] == target:
        return [i,j]

   return None


start = time.time()
result = two_sum_brute(nums, target)
end = time.time()

print('결과:', result)
print(f'걸린 시간: {end - start:.3f}초')
````

- 왜 이렇게 느려질까요? 무엇을 반복해서 하고 있는 걸까요?

## 3️⃣ n을 늘려가며 두 방법 직접 비교하기
n이 커질수록 두 방법의 격차가 얼마나 벌어지는지 눈으로 확인해봅시다.

````python
for size in [1000, 3000, 6000, 10000, 15000]:
    sample = random.sample(range(1, 5_000_000), size)
    tgt = sample[-1] + sample[-2]

    t0 = time.time()
    two_sum_brute(sample, tgt)
    t1 = time.time()

    t2 = time.time()
    two_sum_fast(sample, tgt)
    t3 = time.time()

    print(f'n={size:>6} | 이중반복문: {t1 - t0:7.3f}초 | 딕셔너리 활용: {t3 - t2:8.5f}초')
````

## 🔑 정리

| 방법 | 시간 복잡도 | 공간 복잡도 |
|---|---|---|
| 이중 반복문 (brute force) | O(n²) | O(1) |
| 딕셔너리 활용 (hash) | O(n) | O(n) |

시간을 크게 줄이는 대신, 등장한 값을 저장해두는 딕셔너리 때문에 **추가 공간을 사용**하게 되었습니다.
이것이 **시간-공간 트레이드오프(Time-Space Trade-off)**의 실제 사례입니다.
